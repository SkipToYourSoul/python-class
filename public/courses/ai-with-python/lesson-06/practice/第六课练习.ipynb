## 结课挑战 · 魔王决战综合演示

课堂观看视频与代码同步演示：获取网页 → 保存 CSV → 读取分析 → 画散点图 → 根据证据选择突破口。

魔王情报独立保存为 boss.csv，不能以三塔记录代替。先筛选外罩低于30的部位，再从这些部位中比较防御力，最后按 weakness 选择装备。

以下为教师演示的完整代码。先运行环境格，current_url 会指向本机魔王情报网页。

````python
import requests
from bs4 import BeautifulSoup

res = requests.get(current_url, timeout=10)
res.raise_for_status()
res.encoding = "utf-8"
soup = BeautifulSoup(
    res.text, "html.parser")
csv_text = soup.find(
    id="intel-data").get_text()
with open("boss.csv", "w",
          encoding="utf-8") as f:
    f.write(csv_text)

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

boss = pd.read_csv("boss.csv")
weak = boss[boss["energy"] < 30]
print(weak.sort_values("defense"))

plt.figure(figsize=(9, 5))
sns.scatterplot(data=boss, x="defense",
    y="energy", hue="weakness",
    hue_order=["强光", "火焰", "热量"],
    s=150)
for row in boss.itertuples():
    plt.text(row.defense + 1,
             row.energy - 3, row.part)
plt.axhline(30, color="#F36D4A", ls="--")
plt.ylim(0, 100)
plt.xlim(0, 100)
plt.xlabel("部位防御力")
plt.ylabel("外罩强度")
plt.tight_layout()
plt.savefig("boss-weakness.png", dpi=150)
plt.show()
````

演示输出：boss.csv 有9处部位；候选按防御升序为胸甲、腰甲、左臂。胸甲外罩18、防御24、弱点火焰；生成 boss-weakness.png。
