## 训练并检查成绩

运行下面两个单元格，看模型在 45 张测试卡上答对了多少。

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
svm_classifier = SVC(kernel="linear")
svm_classifier.fit(X_train, y_train)

y_pred = svm_classifier.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"准确率：{accuracy:.2%}")
print(classification_report(
    y_test, y_pred, target_names=iris.target_names
))
```

## 填入你的四项读数

按萼片长、萼片宽、花瓣长、花瓣宽的顺序，把四个 None 改成你测得的数字（厘米）。运行并记下输出的种名。

```python
sepal_length = None
sepal_width = None
petal_length = None
petal_width = None

my_flower_feature = [
    sepal_length, sepal_width,
    petal_length, petal_width,
]
if None in my_flower_feature:
    raise ValueError("请先填入四项测量值（厘米）。")

my_flower_res = svm_classifier.predict(
    [my_flower_feature]
)
name = iris.target_names[my_flower_res[0]]
print("我的鸢尾花属于：", name)
```
