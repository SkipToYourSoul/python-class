### 01 · 读取文件，解析电影资料
在下方一个空白代码单元格中亲手输入完整流程，再按 Shift + Enter 运行。路径从程序当前工作文件夹开始，文件与 Notebook 放在一起。完成标准：片名、导演、评分和两位演员都与样本一致。

完整参考：
```python
from bs4 import BeautifulSoup

path = "./film_1292052.txt"
with open(path, "r", encoding="utf-8") as f:
    html = f.read()

soup = BeautifulSoup(html, "html.parser")
title = soup.find("span", property="v:itemreviewed")
director = soup.find("a", rel="v:directedBy")
score = soup.find("strong", property="v:average")
actors = soup.find_all("a", rel="v:starring")

print("片名：" + title.get_text(strip=True))
print("导演：" + director.get_text(strip=True))
print("评分：" + score.get_text(strip=True))
for actor in actors:
    print("演员：" + actor.get_text(strip=True))
```

运行后核对：
```text

导演：弗兰克·德拉邦特
评分：9.7
演员：蒂姆·罗宾斯
演员：摩根·弗里曼
```

### 参考 · 生成城堡情报表
从上到下亲手输入完整代码并运行，打开生成的 handover.csv。核对三列、一行表头和三条情报，没有多余空行。写文件没有 print，输出区为空是正常的；下面是文件内容示例。csv 模块随 Python 提供，不需要另外安装。

完整参考：
```python
records = [
    {"恶魔": "炎角兽", "地点": "北方峡谷", "弱点": "怕强光"},
    {"恶魔": "藤甲魔", "地点": "迷雾森林", "弱点": "怕火"},
    {"恶魔": "冰翼魔", "地点": "冰封山口", "弱点": "怕热"},
]

import csv

fields = ["恶魔", "地点", "弱点"]
path = "./handover.csv"
with open(path, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fields)
    writer.writeheader()
    writer.writerows(records)
```

文件内容示例：
```text
恶魔,地点,弱点
炎角兽,北方峡谷,怕强光
藤甲魔,迷雾森林,怕火
冰翼魔,冰封山口,怕热
```

### 拓展 · 重新读取表格并核对
