## 第一次请求
url 留空则请求本机第一版；有部署网址后填入完整网址。

```python
import requests
from common import local_sample_server

url = ""
if url.strip():
    res = requests.get(url.strip(), timeout=10)
    print("来源：老师提供的网址")
else:
    with local_sample_server("scout.html") as local_url:
        res = requests.get(local_url, timeout=10)
    print("来源：本机第一版侦察记录")
print("HTTP 状态码：", res.status_code)
res.raise_for_status()
res.encoding = "utf-8"
print(res.text)
```

预期：状态码 200；HTML 包含炎角兽、北方峡谷、火焰、怕强光。
自己核对网页，找到对应证据。

## 下一页再运行：请求第二版
这是预先准备的变化样本，不会后台自动更新。updated_url 留空则实际请求本机第二版。

```python
updated_url = ""
if updated_url.strip():
    res = requests.get(updated_url.strip(), timeout=10)
    print("来源：老师提供的第二版网址")
else:
    with local_sample_server("scout-updated.html") as local_url:
        res = requests.get(local_url, timeout=10)
    print("来源：本机第二版侦察记录")
print("HTTP 状态码：", res.status_code)
res.raise_for_status()
res.encoding = "utf-8"
print(res.text)
```

预期：炎角兽出没地点改为东部石桥，新增岩背魔。
